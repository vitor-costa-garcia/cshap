# Bearing Fault Diagnosis via FFT

Aplicação: Detecção de Falhas em Rolamentos  
Dataset: [Vibration Faults Datasets for Rotating Machines](https://www.kaggle.com/datasets/sumairaziz/vibration-faults-dataset-for-rotating-machines/data)

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import cycshap
np.random.seed(42)
```

```python
# Simular sinais de vibração
fs = 10000
t = np.arange(0, 0.5, 1/fs)

# Gerar sinais com FFT
normal = np.random.randn(100, len(t))
inner_race = normal + 0.5*np.sin(2*np.pi*3500*t)
outer_race = normal + 0.3*np.sin(2*np.pi*1200*t)

all_signals = np.vstack([normal, inner_race, outer_race])
X_fft = np.abs(np.fft.fft(all_signals))[:, :50]
y = np.concatenate([np.zeros(100), np.ones(100), 2*np.ones(100)])

model = RandomForestClassifier(n_estimators=100)
model.fit(X_fft, y)

features = [f'{i*100}Hz' for i in range(10)]
explainer = cycshap.scs(model, X_fft[:, :10])
shap_vals = explainer.shap_values(X_fft[:10, :10])
cycshap.bar_cycmag(shap_vals, feature_names=features[:10])
```
