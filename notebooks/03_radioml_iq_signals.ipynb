# I/Q Signal Classification

Aplicação: Modulação I/Q (Plano Complexo)  
Dataset: [RadioML 2018](https://www.kaggle.com/datasets/pinxau1000/radioml2018)

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import cycshap
np.random.seed(42)
```

```python
# Gerar sinais BPSK e QPSK
X_bpsk = np.random.choice([-1, 1], (200, 10)) + 0.1*np.random.randn(200, 10)
X_qpsk = np.random.choice([-1, 1], (200, 10)) + 1j*np.random.choice([-1, 1], (200, 10)) + 0.15*np.random.randn(200, 10)

X = np.vstack([X_bpsk, X_qpsk])
y = np.concatenate([np.zeros(200), np.ones(200)])

X_mag = np.abs(X)
model = RandomForestClassifier(n_estimators=100)
model.fit(X_mag, y)

explainer = cycshap.scs_complex(model, X_mag)
shap_vals = explainer.shap_values(X_mag[:5])
fig = cycshap.plot_complex_shap_step(shap_vals, X[:5], sample_idx=0)
```
